```python
import os
from dotenv import load_dotenv
import requests

load_dotenv()
BOT_TOKEN = os.getenv('TELEGRAM_BOT_TOKEN')

TELEGRAM_URI = "https://api.telegram.org/bot"

res = requests.get(TELEGRAM_URI + BOT_TOKEN + '/getUpdates')
```
