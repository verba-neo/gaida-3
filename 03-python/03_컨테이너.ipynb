## 리스트 (List)

```python
numbers = [10, 20, 30, 40, 50]
numbers[0]  # 10
numbers[2]  # 30

numbers[-1]  # 50
numbers[-2]  # 40

numbers[0] + numbers[-2]  # 10 + 40

# Index 벗어남 에러
# numbers[5]
# numbers[-6]
```

```python
numbers = [10, 20, 30, 40, 50]

# 슬라이싱 [x:y]  -> x부터 y전까지 을 잘라서 리스트로 반환
numbers[0:3]  # idx 0, 1, 2
numbers[2:2]  # 타겟 idx 없음 -> []

numbers[0:5]  # 0, 1, 2, 3, 4
numbers[:3]   # 처음부터 -> 0 대신 비워도 됨
numbers[2:]   # 끝까지 -> 마지막 비우면 됨

numbers[:]    # 처음부터 끝까지

numbers == numbers[:]  # True -> 들고있는 값이 같음

numbers is numbers[:]  # is 연산자는 존재의 동일함을 비교
```

```python
# 리스트 수정
numbers = [10, 20, 30, 40, 50]

# 특정 index 의 데이터 바꾸기
numbers[2] = 300
numbers[-1] = 500
print(numbers)

# 리스트에 데이터 추가하기 (맨뒤)
numbers.append(60)
print(numbers)

# 리스트의 데이터 삭제하기 (인덱스)
numbers.pop(1)
print(numbers)

list_2d = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0],
]
print(list_2d)

l1 = [1, 2, 3]
l2 = [4, 5, 6]

l1 + l2
```

## 범위(Range)

```python
# range 는 처음부터 마지막-1 까지의 정수만을 의미
r1 = range(1, 10)
# range는 눈에 바로 안보임 -> 리스트로 바꾸면 보임
print(list(r1))

# 시작숫자 x -> 0부터 시작
r2 = range(5)
print(list(r2))

# step 개념
r3 = range(0, 10, 3)
print(list(r3))

# 음수 step(뒤로가기)
r4 = range(10, 0, -1)
print(list(r4))
```

```python
import random

# 대상에서 랜덤하게 하나 뽑아줌
random.choice(['김', '이', '박', '최', '정'])
# 대상에서 랜덤하게 k개 뽑아줌
random.sample(['김', '이', '박', '최', '정'], 2)

x = ['김', '이', '박', '최', '정']
# 랜덤하게 섞어줌
random.shuffle(x)
print(x)
```

```python
# balls 에 1~45 숫자를 저장
balls = range(1, 46)

# random.sample()을 사용해서 6개 랜덤 숫자를 뽑아서, lucky 에 저장
lucky = random.sample(balls, 6)

# 오름차순 정렬
lucky.sort()
# 내림차순 정렬
# lucky.sort(reverse=True)

print(lucky)
```

## Dict (딕셔너리)
| key - Value
- JSON 의 뜻은 JavaScript Object Notation 의 줄임말
    - JS의 Object자료형 식 표기법 (" 써야함)
- Python 에는 Dict 자료형

```python
# 생성
me = {
    'name': 'yu',
    'age': 20,
    'is_hungry': False,
    'height': 190.123,
    'skills': ['n8n', 'sql', 'python']
}
```

```python
# 접근 -> 대괄호 안에 key 를 쓴다.
# 이름, 나이를 뽑아서 출력해 보자
print(me['name'], me['age'])

# 가진 기술 목록을 출력해 보자
print(me['skills'])

# 가진 기술들 중에 'python' 을 꺼내보자.
s = me['skills']  # ['n8n', 'sql', 'python']
s[-1]

# 한번에 접근
me['skills'][-1]
```

```python
# Value 변경
me['name'] = 'kim'

# 배를 고프게 만들고,
me['is_hungry'] = True

# skills 에 'git' 을 추가
me['skills'].append('git')

me
```

```python
from pprint import pprint

# K-V 추가
me['money'] = 100000
pprint(me)

# K-V 제거
me.pop('height')
pprint(me)
```

## Set (집합)
- 중복되는 요소가 없다.
- 순서(index)가 없다.

```python
s1 = {1, 1, 2, 3, 3, 3}
s2 = {3, 4, 5, 5}

print(s1, s2)

# 합집합
print(s1 | s2)
# 교집합
print(s1 & s2)
# 차집합
print(s1 - s2)

list(s1 | s2)

# 리스트에서 중복 없애기 (중복있는 리스트 -> 중복없는 set -> 중복없는 리스트)
x = [1, 1, 1, 1, 1, 1, 1, 2, 2, 2, 3, 2, 2, 3, 2, 2,3,1,3]
uniq_x = list(set(x))
```

## Tuple (튜플)
- 리스트 친척
- 순서(index)가 있음
- 리스트와 차이: 수정 불가능

```python
t = (1, 2, 3, 4)
# 리스트랑 쓰는법 매우 비슷
print(t[0], t[-1], t[1:3])
```
