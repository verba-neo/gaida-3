## 리스트 (List)

```python
numbers = [10, 20, 30, 40, 50]
numbers[0]  # 10
numbers[2]  # 30

numbers[-1]  # 50
numbers[-2]  # 40

numbers[0] + numbers[-2]  # 10 + 40

# Index 벗어남 에러
# numbers[5]
# numbers[-6]
```

```python
numbers = [10, 20, 30, 40, 50]

# 슬라이싱 [x:y]  -> x부터 y전까지 을 잘라서 리스트로 반환
numbers[0:3]  # idx 0, 1, 2
numbers[2:2]  # 타겟 idx 없음 -> []

numbers[0:5]  # 0, 1, 2, 3, 4
numbers[:3]   # 처음부터 -> 0 대신 비워도 됨
numbers[2:]   # 끝까지 -> 마지막 비우면 됨

numbers[:]    # 처음부터 끝까지

numbers == numbers[:]  # True -> 들고있는 값이 같음

numbers is numbers[:]  # is 연산자는 존재의 동일함을 비교
```

```python
# 리스트 수정
numbers = [10, 20, 30, 40, 50]

# 특정 index 의 데이터 바꾸기
numbers[2] = 300
numbers[-1] = 500
print(numbers)

# 리스트에 데이터 추가하기 (맨뒤)
numbers.append(60)
print(numbers)

# 리스트의 데이터 삭제하기 (인덱스)
numbers.pop(1)
print(numbers)
```
