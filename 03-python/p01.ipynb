## 점수 평균과 합격 판별

학생 점수 리스트가 있다.

```py
scores = [80, 75, 90, 60, 100]
```
평균을 구하시오.
평균이 80 이상이면 "합격" 아니면 "불합격" 출력

````python
scores = [80, 75, 90, 60, 100]

# 내장 함수 사용
# avg = sum(scores) / len(scores)

total = 0
for score in scores:
    total += score

avg = total / len(scores)

print(avg)

if avg >= 80:
    print('합격')
else:
    print('불합격')
````

## 최대값 찾기

숫자 리스트가 있다.
```
numbers = [3, 7, 2, 9, 5]
```
가장 큰 값을 찾아 출력하시오.

Tip: `max()` 없이 직접 비교해보기

````python
numbers = [30, 7, 120, 9, 5]

# max(numbers)

# 최댓값 찾기 전의 최댓값 = 첫번째
max_val = numbers[0]

for num in numbers:
    # 비교해서 크면
    if num > max_val:
        # 최댓값 교체
        max_val = num

print(max_val)
````

## 짝수만 저장하기
```
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
```
짝수만 새로운 리스트에 담아서 출력하시오.

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
# 짝수 저장용 새로운 리스트
evens = []

# 하나씩 돌면서
for num in numbers:
    # 짝수면
    if num % 2 == 0:
        # 새 리스트에 추가
        evens.append(num)

print(evens)
````

````python
# List Comprehension
[num for num in numbers if num % 2 == 0]
````

## 이름 존재 여부 확인
```
names = ["철수", "영희", "민수", "지영"]
```
변수 target 에 저장된 이름이 리스트 안에 있으면
"존재합니다" 아니면 "없습니다" 출력

TIP: `in` 연산자 없이 풀기

````python
while True:
    print('무한루프..?')
    break
````

````python
target = '영희'
names = ["철수", "영희", "민수", "지영"]

# 초기 세팅
result = False   # 있으면 True, 없으면 False

# 있나 없나 검사
for name in names:
    if target == name:
        result = True
        break  # 현재 돌고있는 반복문을 종료

if result == True:
    print('존재합니다')
else:
    print('없습니다.')
````
