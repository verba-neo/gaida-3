```python
# 원 (Circle)

# python에서 클래스 이름은 PascalCase (MyGoodMan) , 기존 변수/함수는 snake_case
class Circle:
    pass


# 인스턴스 생성
c1 = Circle()
c2 = Circle()
```

```python
# Person 
# 키, 몸무게, 나이, 이름
# 인사하다, 공부하다

class Person:
    # 1. 특성 (attribute)
    height = 165
    weight = 55
    age = 20
    iq = 100
    name = '인간'

    # 2. 행동 (method)
    def greeting(self):
        return f'안녕하세요, {self.name} 입니다. {self.age}살 입니다.'

    def study(self):
        self.iq += 5

    def exercise(self, time):
        self.weight -= 0.1 * time
        self.age -= 0.01 * time
```

```python
# Person 개념적 대상을 -> 실제 생성
p1 = Person()
p2 = Person()
```

```python
# 기본 속성
print(p1.greeting())
print(p2.greeting())

# 속성 변경
p1.height = 180
p1.name = '김김김'
p1.age = 30
p1.weight = 75
print(p1.greeting())

p2.weight = 60
p2.age = 10
p2.iq = 200
p2.name = '천재'
print(p2.greeting())
```

```python
p1.study()
print(p1.iq)
p1.exercise(10)
print(p1.age, p1.weight)
```

```python
print(p2.iq, p2.name, p2.age, p2.weight)
# 속성 직접 접근/수정
p2.height += 10
# 메서드 실행
p2.study()

p2.exercise(5)
```

```python
# 객체 지향 세상 -> 객체(값)들끼리 상호작용(메서드) 하는 세상
# 모든 코드는 까보면, 객체들의 상호작용 일 뿐이다.

# 일반인용
1 + 2
# 비하인드
(1).__add__(2)

3 > 2
(3).__gt__(2)

[1, 2, 3][0]
[1, 2, 3].__getitem__(0)
```
