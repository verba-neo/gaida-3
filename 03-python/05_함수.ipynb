````python
# name: plus_one / in: 숫자 / out: 숫자 + 1

# 함수 정의(define)
def plus_one(x):
    return x + 1


# 함수 실행/호출(call)
x = plus_one(99)
print(x)
````

````python
# n번 인사하는 함수
# name: hello_ntimes / in: 숫자 / out (return) 없음

# 정의
def hello_ntimes(n):
    # 임시변수를 쓰지 않는 반복문일 경우 임시변수 이름을 _ 로 짓는다.
    for _ in range(n):
        print('반가워')
    # return 을 쓰지 않으면, return None 이 된다.

# 호출
y = hello_ntimes(5)
print(y)
````

````python
# 3제곱 해주는 함수
# name: cube, in: 숫자 1개, out: 숫자의 세제곱
def cube(n):
    return n ** 3

cube(3)  # 27
````

````python
# #
# name: cylinder, in: 반지름 r/높이 h, out: 부피(pi * r^2 * h)
import math

def cylinder(r, h):
    '''원기둥의 부피를 구합니다.
    r -> 반지름
    h -> 높이
    '''
    answer = math.pi * r**2 * h
    return answer


print(cylinder(3, 5))  # 반지름3 높이5인 원기둥의 부피
print(cylinder(5, 3))
````

````python
# 리스트 내부의 숫자들의 평균을 구해주는 함수
# name: get_avg, in: 숫자들로 만들어진 리스트, out: 평균
def get_avg(numbers):
    total = sum(numbers)
    return total / len(numbers)


get_avg([1, 2, 3, 4, 5])  # 3.0
get_avg(list(range(1, 55)))
````

````python
# 비밀번호 체크해주는 함수 (숫자 포함, 8글자 이상)
# name: check_pw, in: 비밀번호 문자열, out: True/False

def check_pw(pw_str):
    is_valid_pw = False

    if len(pw_str) >= 8:
        for char in pw_str:
            if char in '0123456789':
                is_valid_pw = True
                break

    return is_valid_pw



print(check_pw('asdfqwer'))  # F
print(check_pw('1234'))  # F
print(check_pw('asdf1234'))  # T
````

````python
# 비밀번호 체크해주는 함수 (숫자 포함, 8글자 이상, 대문자 포함, 특수문자 포함)
# name: check_pw, in: 비밀번호 문자열, out: True/False


#   매개변수 pw_str이 str 일 거라는 힌트
def check_pw(pw_str: str):
    is_digit = is_upper = is_special = False

    # 함수는 return 을 만나면, 종료
    if len(pw_str) < 8:
        return False

    for char in pw_str:
        # 숫자 체크
        if char.isdigit():
            is_digit = True
        # 대문자 체크
        elif char.isupper():
            is_upper = True
        # 특수문자 체크
        elif char in '!@#$%^&*()_-+':
            is_special = True

    # 3개 다 True일 때만 True가 return 됨
    return is_digit and is_upper and is_special

print(check_pw('asdfqwer'))  # F
print(check_pw('1234'))  # F
print(check_pw('Asdf1234*'))  # T
````
