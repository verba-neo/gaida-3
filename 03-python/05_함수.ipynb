````python
# name: plus_one / in: 숫자 / out: 숫자 + 1

# 함수 정의(define)
def plus_one(x):
    return x + 1


# 함수 실행/호출(call)
plus_one(99)
````

````python
# n번 인사하는 함수
# name: hello_ntimes / in: 숫자 / out (return) 없음


# 정의
def hello_ntimes(n):
    # 임시변수를 쓰지 않는 반복문일 경우 임시변수 이름을 _ 로 짓는다.
    for _ in range(n):
        print('반가워')

# 호출
hello_ntimes(5)
````
