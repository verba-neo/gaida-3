# 04_조건_반복

````python
# 조건 - if
n = -10

if n > 0:
    # 블럭 내부는 조건이 True 일때만 실행됨
    print('양수입니다')
    print('여기는 조건문 블록 안쪽!')

# 블럭 바깥이기 때문에, 조건과 상관 없이 실행됨.
print('조건문 바깥에 있어요')
````

````python
# 조건 - if-else
n = 0

if n > 0:
    print('0보다 크다')
else:
    print('0보다 작거나 같다')
````

````python
# 조건 - if-elif-else
x = 23

if x > 1000:
    print('천넘음')
elif x > 100:
    print('백넘음')
else:
    print('작은수')
````

````python
# 불합: 0-399
# 1: 400-599
# 2: 600-749
# 3: 750-899
# 4: 900-1000
score = 950

if score >= 900:
    lv = 4
elif 900 > score:
    lv = 3
elif 750 > score:
    lv = 2
elif 600 > score:
    lv = 1
else:
    lv = '불합'

print(lv)
````

````python
# 우수 학생 판별
status = {
    'score': 80,
    'absent': 5
}

# 1. 점수가 70을 넘는 학생이면 우수학생
# 2. 우수학생 중에 결석일이 5일 미만이면 최우수학생
# 3. (우수 아닌) 결석일이 5일 미만이면 성실학생
# 4. 나머지는 일반 학생

if status['score'] >= 70 and status['absent'] < 5:
    std = '최우수학생'
elif status['score'] >= 70:
    std = '우수학생'
elif status['absent'] < 5:
    std = '성실학생'
else:
    std = '걍 학생'
    
print(std)

# 중첩 조건문
if status['score'] >= 70:
    if status['absent'] < 5:
        std = '최우수'
    else:
        std = '우수'
elif status['absent'] < 5:
    std = '성실'
else:
    std = '일반'

print(std)

````

## 놀이공원 입장료 계산

놀이공원 입장료를 계산하는 프로그램을 작성하시오.
```py
age = 23
is_weekend = True
has_coupon = False
```
조건은 다음과 같다.

- 기본 입장료는 30,000원
- 13세 미만은 15,000원
- 65세 이상은 20,000원
- 주말에는 기본 입장료에 5,000원 추가
- 쿠폰이 있으면 최종 금액에서 5,000원 할인
-  단, 13세 미만과 65세 이상은 주말 추가 요금이 없다.

최종 입장료를 출력하시오.

````python
age = 25
is_weekend = False # 주말 O/X
has_coupon = True  # 쿠폰 O/X

if age < 13:
    price = 15000
elif age >= 65:
    price = 20000
else:  #  일반인
    price = 30000

    if is_weekend:
        price += 5000

if has_coupon:
    price -= 5000


print(price)
````

## 회원 등급 판정

다음과 같은 고객 정보가 있다.
```py
customer = {
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": True
}
```
고객의 등급을 다음 조건에 따라 판정하시오.

- 회원이 아니면 "일반"
- 회원이면서 구매금액이 500,000원 이상 → "VIP"
- 회원이면서 구매금액이 200,000원 이상 → "GOLD"
- 그 외 회원 → "SILVER"

결과 예시:
```
김민수님의 등급은 GOLD입니다.
```


````python
customer = {
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": True
}

### 1번 방법

if not customer['is_member']:  # False 라면,
    grade = '일반'
# 이하 조건은 모두 회원이 맞음
elif customer['purchase'] >= 500000:
    grade = 'VIP'
elif customer['purchase'] >= 200000:
    grade = 'GOLD'
else:
    grade = 'SILVER'


### 2번 방법
grade = '일반'

if customer['is_member'] and customer['purchase'] >= 500000:
    grade = 'VIP'
elif customer['is_member'] and customer['purchase'] >= 200000:
    grade = 'GOLD'
elif customer['is_member']:
    grade = 'SILVER'


print(f'{customer['name']}님의 등급은 {grade}입니다.')
````

## 로그인 가능 여부 확인

다음 정보를 이용하여 로그인 가능 여부를 판단하시오.
```py
user_id = "admin"
password = "python123"

blocked_users = ["test", "guest", "admin2"]
```
조건:

- 아이디가 `"admin"`이고 비밀번호가 `"python123"`이면 로그인 성공
- 단, `blocked_users`에 아이디가 포함되어 있으면 로그인 실패
- 아이디 또는 비밀번호가 틀리면 `"아이디 또는 비밀번호가 틀렸습니다."`
- 차단된 사용자는 `"차단된 사용자입니다."`

결과를 출력하시오.

````python
user_id = "admin"
password = "python1234"

blocked_users = ["test", "guest", "admin2"]


if user_id in blocked_users:
    print('차단된 사용자입니다.')
elif user_id == 'admin' and password == 'python123':
    print('로그인 성공')
else:
    print('틀림')
````

## 배송비 계산

온라인 쇼핑몰의 주문 정보가 다음과 같다.
```py
order = {
    "price": 48000,
    "is_member": True,
    "region": "서울",
    "coupon": "WELCOME"
}
```
배송비를 계산하시오.

조건:

- 상품 금액이 50,000원 이상이면 무료배송
- 회원이면 배송비 2,000원
- 비회원이면 배송비 3,000원
- `"제주"` 지역이면 위 배송비에 3,000원 추가
- "WELCOME" 쿠폰을 가지고 있으면 배송비 무료
- 단, 상품 금액이 10,000원 미만이면 쿠폰이 있어도 배송비 무료가 아니다.

````python
order = {
    "price": 438000,
    "is_member": False,
    "region": "제주",
    "coupon": "WELCdOME"
}

# 기본 배송비
fee = 3000

# 1. 회원 체크
if order['is_member']:
    fee = 2000

# 2. 오만원이상
if order['price'] >= 50000:
    fee = 0

# 3. 쿠폰 + 만원이상
if order['coupon'] == 'WELCOME' and order['price'] >= 10000:
    fee = 0

# 4. 제주는 별도
if order['region'] == '제주':
    fee += 3000

print(fee)
````

### `while` 구문

````python
n = 0

while n < 5:
    print('출력', n)
    n += 1

print('--종료 이후 n', n)
````

````python
numbers = [1, 2, 3, 4, 5, 6, 800]

total = 0
idx = 0

# idx 가 numbers의 길이(len())보다 작다면, 아래 블럭 반복
while idx < len(numbers):
    # total 에 numbers의 요소를 idx 접근하여 하나씩 추가로 더함
    total += numbers[idx]
    print('중간점검', total)
    # idx 를 1 올린다
    idx += 1

print(total)

# 단순 숫자 더하기면, sum()
sum(numbers)
````

### `for .. in` 구문

````python
numbers = [1, 2, 3, 4, 5]
total = 0

for num in numbers:
    print(f'기존 {total} 에 {num}를 더합니다')
    total += num

print(total)
````

````python
# 1부터 10000까지, 3의 배수 를 다 더해보자

# 1~10까지 3배수 더하기를 먼저하고, 18이 나오면, 10 -> 10000 으로 바꾸자

# Hint
# while 방법
total = 0
i = 3

while i <= 1e4:  # 1 * 10**4
    total += i
    i += 3

print(total)
````

````python
# for (range)
N = 10000

# 1. * 3
# 0 ~ (목표값을 3으로 나눈 몫 + 1) 까지의 3의 배수를 다 더한다
total = 0
for n in range(N // 3 + 1):
    total += n*3
print(total)

# 2. range 의 step 을 활용
total = 0
for n in range(0, N+1, 3):
    total += n

print(total)
````

````python
# 리스트는 idx 수정 가능, 문자열은 idx 수정 안됨
print('리스트 + for')
for l in [1, 2, 3]:
    print(l)

print('레인지 + for')
for r in range(-3, 3):
    print(r)

print('문자열 + for')
for s in 'python':
    print(f'!{s}!')
````
