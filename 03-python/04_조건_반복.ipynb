# 04_조건_반복

````python
# 조건 - if
n = -10

if n > 0:
    # 블럭 내부는 조건이 True 일때만 실행됨
    print('양수입니다')
    print('여기는 조건문 블록 안쪽!')

# 블럭 바깥이기 때문에, 조건과 상관 없이 실행됨.
print('조건문 바깥에 있어요')
````

````python
# 조건 - if-else
n = 0

if n > 0:
    print('0보다 크다')
else:
    print('0보다 작거나 같다')
````

````python
# 조건 - if-elif-else
x = 23

if x > 1000:
    print('천넘음')
elif x > 100:
    print('백넘음')
else:
    print('작은수')
````

````python
# 불합: 0-399
# 1: 400-599
# 2: 600-749
# 3: 750-899
# 4: 900-1000
score = 950

if score >= 900:
    lv = 4
elif 900 > score:
    lv = 3
elif 750 > score:
    lv = 2
elif 600 > score:
    lv = 1
else:
    lv = '불합'

print(lv)
````

````python
# 우수 학생 판별
status = {
    'score': 80,
    'absent': 5
}

# 1. 점수가 70을 넘는 학생이면 우수학생
# 2. 우수학생 중에 결석일이 5일 미만이면 최우수학생
# 3. (우수 아닌) 결석일이 5일 미만이면 성실학생
# 4. 나머지는 일반 학생

if status['score'] >= 70 and status['absent'] < 5:
    std = '최우수학생'
elif status['score'] >= 70:
    std = '우수학생'
elif status['absent'] < 5:
    std = '성실학생'
else:
    std = '걍 학생'
    
print(std)

# 중첩 조건문
if status['score'] >= 70:
    if status['absent'] < 5:
        std = '최우수'
    else:
        std = '우수'
elif status['absent'] < 5:
    std = '성실'
else:
    std = '일반'

print(std)

````
