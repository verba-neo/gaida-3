## 1. 짝수만 골라내는 함수

숫자 리스트를 전달받아 짝수만 새로운 리스트로 만들어 반환하는 `get_even_numbers()` 함수를 작성하시오.

````python
numbers = [12, 7, 4, 19, 22, 31, 8, 15]

# 여기에 함수를 작성하세요.
def get_even_numbers(numbers):  # 매개변수 이름은 함수 바깥 변수 이름과 무관
    evens = []
    for num in numbers:
        if num % 2 == 0:
            evens.append(num)

    evens.sort() # 오름차순 정렬
    return evens

result = get_even_numbers(numbers)

print(result)
# 예상 결과: [12, 4, 22, 8]
````

## 2. 학생 성적 분석 함수

학생들의 정보가 다음과 같이 주어져 있다.
```
students = [
    {"name": "철수", "score": 85},
    {"name": "영희", "score": 92},
    {"name": "민수", "score": 67},
    {"name": "지영", "score": 78},
    {"name": "현우", "score": 95}
]
```
`analyze_students()` 함수를 작성하시오.

함수는 학생 데이터를 받아 다음 정보를 딕셔너리로 반환해야 한다.
```
{
    "average": 83.4,
    "highest": "현우",
    "passed": ["철수", "영희", "지영", "현우"]
}
```

조건:

- 평균 점수 계산
- 가장 높은 점수를 받은 학생 이름 찾기
- 70점 이상인 학생 이름을 passed 리스트에 저장

````python
students = [
    {"name": "철수", "score": 85},
    {"name": "영희", "score": 92},
    {"name": "민수", "score": 67},
    {"name": "지영", "score": 78},
    {"name": "현우", "score": 95}
]


# 여기에 함수를 작성하세요.      # input으로 dict로 이루어진 list가 들어올 것! 이라는 hint
def analyze_students(students: list[dict]):
    # 평균 구하기
    total = 0
    hst = students[0]
    passed = []

    for std in students:
        # 총점 구하기
        total += std['score']
        # 최고점자 갱신하기
        if std['score'] > hst['score']:
            hst = std
        # 70점 체크
        if std['score'] >= 70:
            passed.append(std['name'])

    avg = total / len(students)
    name = hst['name']

    return {'average': avg, 'highest': name, 'passed': passed}


result = analyze_students(students)
result
# {
#     'average': 83.4,
#     'highest': '현우',
#     'passed': ['철수', '영희', '지영', '현우']
# }
````

## 3. 상품 가격 계산기

다음과 같은 상품 목록이 있다.
```
products = [
    {"name": "노트북", "price": 1200000, "quantity": 1},
    {"name": "마우스", "price": 30000, "quantity": 2},
    {"name": "키보드", "price": 80000, "quantity": 1},
    {"name": "모니터", "price": 350000, "quantity": 2}
]
```
상품 목록을 전달받아 전체 주문 금액을 반환하는 `calculate_total()` 함수를 작성하시오.

조건
- 상품별 금액은 price × quantity로 계산한다.
- 상품별 금액이 50만원 이상이면 해당 상품에 5% 할인을 적용한다.
- 할인 적용 후 모든 상품의 금액을 합산한다.
- 최종 금액을 반환한다.

````python
products = [
    {"name": "노트북", "price": 1200000, "quantity": 1},
    {"name": "마우스", "price": 30000, "quantity": 2},
    {"name": "키보드", "price": 80000, "quantity": 1},
    {"name": "모니터", "price": 350000, "quantity": 2}
]

# 여기에 함수를 작성하세요.
def calculate_total(products: list[dict]):
    total = 0
    for p in products:
        # 개별 상품의 가격
        p_price = p['price']
        # 개별 상품의 개수
        p_qty = p['qty']
        # 개별 상품의 총 가격
        p_total = p_price * p_qty
        # 할인조건 체크
        if p_total >= 5e5:
            # 5% 할인
            p_total *= 0.95
        # 더하기
        total += p_total
    return total

result = calculate_total(products)

print(result)  # 1945000
````

## 4. 매출 데이터 분석 ⭐

다음과 같은 매출 데이터가 있다.
```
sales = [
    {"product": "노트북", "category": "전자제품", "amount": 1200000},
    {"product": "마우스", "category": "전자제품", "amount": 30000},
    {"product": "의자", "category": "가구", "amount": 250000},
    {"product": "책상", "category": "가구", "amount": 400000},
    {"product": "모니터", "category": "전자제품", "amount": 350000}
]
```
매출 데이터를 전달받아 다음과 같은 결과를 반환하는 analyze_sales() 함수를 작성하시오.
```
{
    "total": 2230000,
    "electronics": 1580000,
    "furniture": 650000,
    "high_sales": ["노트북", "모니터", "책상"]
}
```
조건
- 전체 매출액을 계산한다.
- "전자제품" 카테고리의 매출액을 계산한다.
- "가구" 카테고리의 매출액을 계산한다.
- 판매 금액이 30만원 이상인 상품명을 `high_sales` 리스트에 저장한다.
- 최종 결과를 하나의 딕셔너리로 반환한다.

````python
sales = [
    {"product": "노트북", "category": "전자제품", "amount": 1200000},
    {"product": "마우스", "category": "전자제품", "amount": 30000},
    {"product": "의자", "category": "가구", "amount": 250000},
    {"product": "책상", "category": "가구", "amount": 400000},
    {"product": "모니터", "category": "전자제품", "amount": 350000}
]

# 여기에 함수를 작성하세요.
def analyze_sales(sales):
    # 초기화
    total = electronics = furniture = 0
    high_sales = []

    for sale in sales:
        # dict 에서 각 항목별로 추출
        product = sale["product"]
        category = sale["category"]
        amount = sale["amount"]
        # 총액 더하기
        total += amount
        # 카테고리별로 더하기
        if category == "전자제품":
            electronics += amount
        elif category == "가구":
            furniture += amount
        # 고액상품 이름 추가
        if amount >= 300000:
            high_sales.append(product)

    return {
        "total": total,
        "electronics": electronics,
        "furniture": furniture,
        "high_sales": high_sales
    }


result = analyze_sales(sales)

print(result)
# {
#     'total': 2230000,
#     'electronics': 1580000,
#     'furniture': 650000,
#     'high_sales': ['노트북', '책상', '모니터']
# }
````

## 5. 상품 재고 관리 함수 ⭐⭐

다음과 같은 상품 데이터가 있다.
```
products = [
    {"name": "노트북", "price": 1200000, "stock": 3},
    {"name": "마우스", "price": 30000, "stock": 0},
    {"name": "키보드", "price": 80000, "stock": 5},
    {"name": "모니터", "price": 350000, "stock": 1},
    {"name": "헤드셋", "price": 70000, "stock": 2}
]
```
상품 데이터를 전달받아 재고 상태를 분석하는 `check_inventory()` 함수를 작성하시오.

조건
- stock이 0이면 "품절"
- stock이 1이면 "재고 부족"
- stock이 2 이상이면 "판매 가능"
- 각 상품의 이름과 재고 상태를 새로운 딕셔너리에 저장한다.
- 최종적으로 상품명과 상태를 담은 딕셔너리를 반환한다.

````python
products = [
    {"name": "노트북", "price": 1200000, "stock": 3},
    {"name": "마우스", "price": 30000, "stock": 0},
    {"name": "키보드", "price": 80000, "stock": 5},
    {"name": "모니터", "price": 350000, "stock": 1},
    {"name": "헤드셋", "price": 70000, "stock": 2}
]
# 여기에 함수를 작성하세요.

def check_inventory(products):
    # HINT: 함수 안에서 다음과 같은 형태로 결과를 만들어야 한다.
    result = {}

    for product in products:
        # 이름, 재고 추출
        name = product["name"]
        stock = product["stock"]

        # name을 key 로 하고, stock에 맞게 value를 설정
        # * dict에 새로운 k-v 추가하는법 => dict[새로운key] = 새로운value
        if stock == 0:
            result[name] = "품절"
        elif stock == 1:
            result[name] = "재고 부족"
        else:
            result[name] = "판매 가능"

    return result


result = check_inventory(products)

print(result)

# {
#     '노트북': '판매 가능',
#     '마우스': '품절',
#     '키보드': '판매 가능',
#     '모니터': '재고 부족',
#     '헤드셋': '판매 가능'
# }
````
