## 비교 연산자
|연산자|내용|
|---|---|
|`>`, `<`|초과, 미만|
|`>=`, `<=`|이상, 이하|
|`==`|값이 같음|
|`!=`|다르다|
|`in`|포함여부|
|`is`|동일하다|

```python
# 크고 작음
5 > 3  # T
4 < 2  # F
4 >= 4  # T
3 <= 1  # F
```

```python
5 == 5  # T
4 != 2  # T
'asdf' == 'qwer'  # F
```

```python
# x in y -> y안에 x가 있는지

'i' in 'apple'  # F
3 in [1, 2, 3, 4]  # T

# 욕설 필터링
msg = '야 이 바보야 겜 접어라'

'바보' in msg  # 욕설필터링
'멍청이' in msg 
```

```python
'ab' < 'ac'  # 글자 => 숫자로 변환

# 글자의 번호, 번호의 글자
ord('나'), chr(33)
```

## 논리 연산자
|연산자|내용|
|---|---|
|`and`|`a and b` -> a, b 모두 True일 경우에만 True|
|`or`|`a or b` -> a, b 모두 False일 경우에만 False|
|`not`|`not True` -> 값을 뒤집는다|

```python
True and True  # T
True and False  # F
False and True  # F
False and False  # F
```

```python
True or True  # T
True or False  # T
False or True  # T
False or False  # F
```

```python
not True, not False

is_married = True

# 사용자가 미혼이라면
if not is_married:
    pass

# 'abc' 안에 char 가 없다면
char = 'a'
char not in 'abc'
```
