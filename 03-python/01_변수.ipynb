```python
# 문제 
# x = 10, y = 20 일때, 값을 서로 교체하는 코드를 작성해 보자.
x = 10
y = 20
print(x, y)

# 아래 코드에는 숫자가 등장하면 안됨!
tmp = x  # tmp 10 / x 10 / y 20
x = y    # tmp 10 / x 20 / y 20
y = tmp  # tmp 10 / x 20 / y 10
print(x, y)

x, y = y, x
print(x, y)

# tmp 없이, swap 안쓰고 바꾸기
x, y = 10, 20

x = x + y  # x 30, y 20
y = x - y  # x 30, y 10
x = x - y  # x 20, y 10
```
